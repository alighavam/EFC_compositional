## Making TGT (chords per subject from participants.csv)

```python
import os
import pandas as pd
import numpy as np

# 4-fing: 12912, 21921, 21912, 12921
# target_chord: the 4-finger chord whose radial and ulnar 2-finger components are trained
# untrained_chords: the other three 4-finger chords
sn = 105
participants = pd.read_csv('participants.csv', dtype=str).set_index('participant_id')
info = participants.loc[f's{sn}']
target_chord = int(info['target_chord'])
trained_chords = [int(c) for c in info['trained_components'].split(';')]
untrained_chords = [int(c) for c in info['untrained_chords'].split(';')]
assert trained_chords == [int(str(target_chord)[:2] + '999'), int('999' + str(target_chord)[3:])]
test_chords = [target_chord] + untrained_chords # all four 4-finger chords

# servDir = '/Volumes/Diedrichsen_data$/data/Chord_exp/EFC_EMG_compositional' # save if server is available
targetDir = 'target' # server and local directory for target files

cols = ['subNum', 'chordID', 'planTime', 'success_holdTime', 'execMaxTime', 'feedbackTime', 'iti']
planTime = 1000
success_holdTime = 600
execAccTime = 1000
execMaxTime = 10000
feedbackTime = 1000
iti = 500

# TESTING SESSION 1:
repetitions = 2
nchuncks = 10
nRuns = 4
chordID = test_chords
chordID = np.repeat(chordID, nchuncks)
np.random.shuffle(chordID)
chordID = np.repeat(chordID, repetitions)

# break into runs:
for i in range(nRuns):
    df_run = pd.DataFrame(columns=cols)
    ntrials_perrun = len(chordID)//nRuns

    df_run['subNum'] = [sn]*ntrials_perrun
    df_run['chordID'] = chordID[i*ntrials_perrun:(i+1)*ntrials_perrun]
    df_run['planTime'] = planTime*np.ones(ntrials_perrun) + np.random.randint(100, 400, ntrials_perrun) # random jitter
    df_run['success_holdTime'] = success_holdTime*np.ones(ntrials_perrun)
    df_run['execMaxTime'] = execMaxTime*np.ones(ntrials_perrun)
    df_run['feedbackTime'] = feedbackTime*np.ones(ntrials_perrun)
    df_run['iti'] = iti*np.ones(ntrials_perrun)

    df_run = df_run.astype(int)
    
    df_run.to_csv(os.path.join(f'./{targetDir}', f's{sn}_day1_{i+1:02d}.tgt'), index=False, sep='\t')

# Training Session 1-3:
repetitions = 2
nchuncks = 50
nRuns = 10
for day in range(1, 4):
    chordID = trained_chords
    chordID = np.repeat(chordID, nchuncks)
    np.random.shuffle(chordID)
    chordID = np.repeat(chordID, repetitions)
    
    # break into runs:
    for i in range(nRuns):
        df_run = pd.DataFrame(columns=cols)
        ntrials_perrun = len(chordID)//nRuns

        df_run['subNum'] = [sn]*ntrials_perrun
        df_run['chordID'] = chordID[i*ntrials_perrun:(i+1)*ntrials_perrun]
        df_run['planTime'] = planTime*np.ones(ntrials_perrun) + np.random.randint(100, 400, ntrials_perrun) # random jitter
        df_run['success_holdTime'] = success_holdTime*np.ones(ntrials_perrun)
        df_run['execMaxTime'] = execMaxTime*np.ones(ntrials_perrun)
        df_run['feedbackTime'] = feedbackTime*np.ones(ntrials_perrun)
        df_run['iti'] = iti*np.ones(ntrials_perrun)

        df_run = df_run.astype(int)

        df_run.to_csv(os.path.join(f'./{targetDir}', f's{sn}_day{day}_{(day-1)*nRuns+i+5:02d}.tgt'), index=False, sep='\t')

# Testing Session 2:
repetitions = 2
nchuncks = 20
nRuns = 8
chordID = test_chords
chordID = np.repeat(chordID, nchuncks)
np.random.shuffle(chordID)
chordID = np.repeat(chordID, repetitions)

# break into runs:
for i in range(nRuns):
    df_run = pd.DataFrame(columns=cols)
    ntrials_perrun = len(chordID)//nRuns

    df_run['subNum'] = [sn]*ntrials_perrun
    df_run['chordID'] = chordID[i*ntrials_perrun:(i+1)*ntrials_perrun]
    df_run['planTime'] = planTime*np.ones(ntrials_perrun) + np.random.randint(100, 400, ntrials_perrun) # random jitter
    df_run['success_holdTime'] = success_holdTime*np.ones(ntrials_perrun)
    df_run['execMaxTime'] = execMaxTime*np.ones(ntrials_perrun)
    df_run['feedbackTime'] = feedbackTime*np.ones(ntrials_perrun)
    df_run['iti'] = iti*np.ones(ntrials_perrun)
    df_run = df_run.astype(int)
    df_run.to_csv(os.path.join(f'./{targetDir}', f's{sn}_day3_{i+1+34:02d}.tgt'), index=False, sep='\t')

# Post-training:
repetitions = 2
nchuncks = 25
nRuns = 10
chordID = test_chords
chordID = np.repeat(chordID, nchuncks)
np.random.shuffle(chordID)
chordID = np.repeat(chordID, repetitions)

# break into runs:
for i in range(nRuns):
    df_run = pd.DataFrame(columns=cols)
    ntrials_perrun = len(chordID)//nRuns

    df_run['subNum'] = [sn]*ntrials_perrun
    df_run['chordID'] = chordID[i*ntrials_perrun:(i+1)*ntrials_perrun]
    df_run['planTime'] = planTime*np.ones(ntrials_perrun) + np.random.randint(100, 400, ntrials_perrun) # random jitter
    df_run['success_holdTime'] = success_holdTime*np.ones(ntrials_perrun)
    df_run['execMaxTime'] = execMaxTime*np.ones(ntrials_perrun)
    df_run['feedbackTime'] = feedbackTime*np.ones(ntrials_perrun)
    df_run['iti'] = iti*np.ones(ntrials_perrun)
    df_run = df_run.astype(int)
    df_run.to_csv(os.path.join(f'./{targetDir}', f's{sn}_day4_{i+1+42:02d}.tgt'), index=False, sep='\t')

```
